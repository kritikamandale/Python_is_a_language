```python
import seaborn as sns
```

```python
data = sns.load_dataset('titanic')
```

```python
data.head()
```

```python
data.info()
```

```python
data.sample(10)
```

```python
data.drop("embarked", axis=1,inplace = True)
```

```python
data.head()
```

```python
data.columns
```

```python
data.drop(['who', 'adult_male', 'deck', 'embark_town', 'alive', 'alone'],axis = 1,inplace=True)
```

```python
data.head()
```

```python
data.isna().sum()
```

```python
data['age'].fillna(data['age'].mean(), inplace = True)
```

```python
data.isna().sum()
```

```python
data['age'] = data['age'].astype(int)
```

```python
import pandas as pd
```

```python
data_encoded = pd.get_dummies(columns=['sex'],data=data,dtype = int, drop_first = True)
```

```python
data_encoded.head()
```

```python
data_encoded.to_csv("titanic_cleaned.csv", index = False)
```
