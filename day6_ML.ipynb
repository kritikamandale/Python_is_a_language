```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('environmental_socioeconomic.csv')
```

```python
data
```

```python
data.info()
```

```python
data.columns
```

```python
from sklearn.preprocessing import StandardScaler
#normalize the feature using standardscaler
scalar = StandardScaler()
data_scaled = scalar.fit_transform(data)
```

```python
import scipy.cluster.hierarchy as sch
link = sch.linkage(data_scaled, method='ward')
dend = sch.dendrogram(link)
```

```python
from sklearn.cluster import AgglomerativeClustering
model = AgglomerativeClustering(n_clusters=3, metric = 'euclidean', linkage='ward')
labels = model.fit_predict(data_scaled)
labels
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.scatter(data_scaled[:,0],data_scaled[:,1], c = model.labels_, cmap = 'rainbow')
plt.show()
```

```python
from sklearn.datasets import load_breast_cancer
cancer = load_breast_cancer()
cancer
```

```python
type(cancer)
```

```python
print(cancer['DESCR'])
```

```python
#creating the dataframe from this bunch data
cancer_df = pd.DataFrame(cancer['data'], columns = cancer['feature_names'])
cancer_df.head()
```

```python
cancer_df.info()
```

```python
from sklearn.preprocessing import StandardScaler
#normalize the feature using standardscaler
scalar = StandardScaler()
cancer_df_scaled = scalar.fit_transform(cancer_df)
cancer_df_scaled[:5]
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
x_new = pca.fit_transform(cancer_df_scaled)
x_new.shape
```

```python
cancer_df_scaled.shape
```

```python
import matplotlib.pyplot as plt
plt.scatter(x_new[:,0], x_new[:,1], c = cancer['target'], cmap = 'plasma')
plt.show()
```

```python
pca.explained_variance_
```
